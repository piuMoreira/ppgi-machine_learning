```python
import numpy as np
import pandas as pd
from skimage.feature import graycomatrix, graycoprops
from skimage.measure import shannon_entropy
import matplotlib.pyplot as plt
import os
import cv2

# Example of how to read and show image from the data set in grayscale
DATADIR = 'Data/DeFungi-H1'

for img in os.listdir(DATADIR):
    img_array = cv2.imread(os.path.join(DATADIR,img), cv2.IMREAD_GRAYSCALE)
    plt.imshow(img_array, cmap='gray')
    plt.show()
    break
```

```python
# The data set used is already sanitazed, so all images have the same dimentions
# If there were need to resize the images this would be done:

IMG_SIZE = 500

new_array = cv2.resize(img_array, (IMG_SIZE,IMG_SIZE))
plt.imshow(new_array, cmap='gray')
plt.show()
```

```python
# Reading all images and extrating data
DATADIR = 'Data/DeFungi-H1'

img_array = []

for file in os.listdir(DATADIR):
    img = cv2.imread(os.path.join(DATADIR,file), cv2.IMREAD_GRAYSCALE)
    # Entropy
    entropy = shannon_entropy(img)
    # Edges
    edges = cv2.Canny(img, 100, 200)
    # Edge density
    edge_density = np.sum(edges > 0) / edges.size
    # Texture
    glcm = graycomatrix(
        img,
        distances=[1],
        angles=[0],
        levels=256,
        symmetric=True,
        normed=True
    )
    # contrast
    contrast = graycoprops(glcm, 'contrast')[0, 0]
    # Homogeneity
    homogeneity = graycoprops(glcm, 'homogeneity')[0, 0]

    img_array.append([file,entropy,edge_density,contrast,homogeneity])
```

```python
# Create DataFrame
df = pd.DataFrame(img_array, columns=['file','entropy','edge_density','contrast','homogeneity'])

# Save to CSV
df.to_csv("image_dataCharacterization.csv", index=False)

print(df.head())
```
